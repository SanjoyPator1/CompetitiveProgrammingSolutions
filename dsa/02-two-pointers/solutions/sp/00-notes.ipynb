## Types of Two Pointers

### PATTERN 1: OPPOSITE DIRECTION (CONVERGING POINTERS)

**Philosophy**: Start from extremes and move toward center based on conditions.

**When to use**: This is your go-to pattern when you have a **sorted array** or when you need to **compare elements from both ends**. The key insight is that you can eliminate large portions of the search space with each comparison.

**How it works**:

- Place one pointer at the start (left = 0) and one at the end (right = length - 1)
- Compare or calculate something with the values at both pointers
- Based on the result, move the appropriate pointer inward
- Continue until pointers meet or cross

**Real-world analogy**: Think of it like two people walking toward each other from opposite ends of a hallway. They stop and make decisions at each step based on what they're looking for.

```python
import time
import random
from typing import List, Optional, Tuple
from collections import Counter, defaultdict
```

#### Example 1: Two Sum

```python
def two_sum_sorted(nums: List[int], target: int) -> List[int]:
    """Find two numbers in sorted array that add up to target.
    Time: O(n), Space: O(1)

    Args:
        nums (List[int]): _description_
        target (int): _description_

    Returns:
        List[int]: _description_
    """
    left, right = 0, len(nums)-1

    while left<right:
        current_sum = nums[left] + nums[right]
        print(f"Left: {left}({nums[left]}), Right: {right}({nums[right]}), Sum: {current_sum}")
        
        if current_sum == target:
            return [left,right]
        elif current_sum < target:
            left += 1
        else:
            right -=1
    
    return []
```

```python
# Test
print("=== TWO SUM SORTED ===")
test_arr = [2, 7, 11, 15]
target = 9
result = two_sum_sorted(test_arr, target)
print(f"Array: {test_arr}, Target: {target}")
print(f"Result: {result}")
print()
```

#### Example 2: Palindrome

```python
def is_palindrome_verbose(s: str) -> bool:
    """Check if string is palindrome with detailed logging.

    Args:
        s (str): _description_

    Returns:
        bool: _description_
    """
    left, right = 0, len(s)-1

    while left < right:
        print(f"Comparing s[{left}]='{s[left]}' with s[{right}]='{s[right]}'")
        
        if s[left] != s[right]:
            print("Mismatch found")
            return False
        
        left += 1
        right -= 1

    return True
```

```python
# Test Palindrome
print("=== PALINDROME CHECK ===")
test_strings = ["racecar", "hello", "a", ""]
for s in test_strings:
    print(f"Testing '{s}':")
    result = is_palindrome_verbose(s)
    print(f"Is palindrome: {result}\n")
```

#### Example 3: Three Sum

```python
def three_sum_detailed(nums: List[int]) ->  List[List[int]]:
    """Find all unique triplets that sum to zero with detailed logging.

    Args:
        nums (List[int]): _description_

    Returns:
        List[List[int]]:: _description_
    """
    # first sort the numbers so that we can implement two pointers
    nums.sort()
    result = []
    print(f"Sorted array: {nums}")
    print("running for loop for i=",len(nums)-2)

    for i in range(len(nums)-2):
        # skip duplicates for first element
        if i>0 and nums[i] == nums[i-1]:
            print(f"Skipping duplicate at index {i}: {nums[i]}")
            continue

        left, right = i+1, len(nums) - 1
        # we want nums[left] + nums[right] = -nums[i]
        target = -nums[i]
        print(f"\nFixed nums[{i}] = {nums[i]}, looking for two numbers that sum to {target}")

        while left<right:
            current_sum = nums[left] + nums[right]
            print(f"  Left: {left}({nums[left]}), Right: {right}({nums[right]}), Sum: {current_sum}")

            if current_sum == target:
                triplet = [nums[i], nums[left], nums[right]]
                result.append(triplet)
                print(f"  Found triplet: {triplet}")

                # skip duplicates
                while left<right and nums[left] == nums[left+1]:
                    left += 1
                while left<right and nums[right] == nums[right-1]:
                    right -= 1

                left += 1
                right -= 1
            elif current_sum < target:
                left += 1
            else:
                right -= 1
    
    return result
```

```python
# Test Three Sum
print("=== THREE SUM ===")
test_nums = [-1, 0, 1, 2, -1, -4]
result = three_sum_detailed(test_nums)
print(f"Final result: {result}")
print()
```

#### Example 4: Valid Palindrome (Skip Non-Alphanumeric)

```python
def is_valid_palindrome(s):
    """Check palindrome considering only alphanumeric characters.

    Example: "A man, a plan, a canal: Panama"
    Becomes: "AmanaplanacanalPanama" → "amanaplanacanalpanama"

    The algorithm skips spaces, commas, colons while comparing.
    Time: O(n), Space: O(1)

    Args:
        s (_type_): _description_
    """
    left, right = 0, len(s)-1

    while left<right:
        # skip non-alphanumeric from left
        while left<right and not s[left].isalnum():
            left += 1

        # skip non-alphanumeric from right
        while left<right and not s[right].isalnum():
            right -= 1

        # compare left and right (case insensitive)
        if s[left].lower() != s[right].lower():
            return False
        
        left += 1
        right -= 1

    return True
```

```python
# Test Palindrome
print("=== PALINDROME CHECK skip non alphanumeric ===")
test_strings = ["A man, a plan, a canal: Panama", "hello", "a", ""]
for s in test_strings:
    print(f"Testing '{s}':")
    result = is_valid_palindrome(s)
    print(f"Is palindrome: {result}\n")
```

### PATTERN 2: SAME DIRECTION (READ/WRITE POINTERS)

**Philosophy**: One pointer explores ahead while another follows, often at different speeds.

**When to use**: This pattern is perfect for **in-place array modifications**. Use it when you need to remove elements, filter data, or rearrange arrays without using extra space.

**How it works**:

- Both pointers start at the beginning (or specific positions)
- The "fast" pointer reads/explores elements
- The "slow" pointer writes/keeps track of where valid elements should go
- The fast pointer moves every iteration, slow pointer moves only when needed

**Real-world analogy**: Think of it like two people cleaning a messy room. One person (fast pointer) looks at every item, while the other person (slow pointer) only moves when they find something worth keeping, placing it in the "clean" section.


#### Example 1: Remove duplicates from sorted array

```python
def remove_duplicates_verbose(nums: List[int]) -> int:
    """Remove duplicates from sorted array with detailed logging.

    Args:
        nums (List[int]): _description_

    Returns:
        int: _description_
    """

    if not nums:
        return 0
    
    write_idx = 1 # First element always unique

    print(f"Original array: {nums}")

    for read_idx in range(1, len(nums)):
        print(f"Reading nums[{read_idx}]={nums[read_idx]}")
        
        # if the read index found a new number put it in place of the write index
        if nums[read_idx] != nums[read_idx-1]:
            print(f"  Unique! Writing to position {write_idx}")
            nums[write_idx] = nums[read_idx]
            # increment write index
            write_idx += 1
        else:
            print(f"  Duplicate of nums[{read_idx-1}]={nums[read_idx-1]}, skipping")

        print(f"  Array state: {nums[:write_idx]} + {nums[write_idx:]}")

    return write_idx
```

```python
# Test Remove Duplicates
print("=== REMOVE DUPLICATES ===")
test_nums = [1, 1, 2, 2, 3, 4, 4, 5]
new_length = remove_duplicates_verbose(test_nums.copy())
print(f"New length: {new_length}")
print()
```

#### Example 2: Move zeroes to End

Problem Setup: Move all zeros to the end while maintaining the relative order of non-zero elements.

Strategy: Use two passes - first move all non-zero elements to the front, then fill the rest with zeros.

```python
def move_zeros_verbose(nums: List[int]) -> None:
    """
    Move all zeros to end while maintaining relative order.

    Time: O(n), Space: O(1)
    """
    write_idx = 0

    print(f"Original array: {nums}")
    print("Phase 1: Move non-zero elements to front")

    # first move all non zero to the front
    for read_idx in range(len(nums)):
        if nums[read_idx] != 0:
            print(f"  Moving nums[{read_idx}]={nums[read_idx]} to position {write_idx}")
            nums[write_idx] = nums[read_idx]
            write_idx += 1
        else:
            print(f"  Skipping zero at index {read_idx}")

        print(f"  Array state: {nums} \n")

    print("Phase 2: Fill remaining positions with zeros")
    while write_idx < len(nums):
        nums[write_idx] = 0
        write_idx += 1

    print(f"Final array: {nums}")
```

```python
# Test Move Zeros
print("=== MOVE ZEROS ===")
test_nums = [0, 1, 0, 3, 12]
move_zeros_verbose(test_nums)
print()
```

### PATTERN 3: FAST-SLOW POINTERS

#### Example 1: Cycle Detection

```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

    def __repr__(self):
        return f"Node({self.val})"
    
def create_linked_list(values: List[int], cycle_pos: int = -1) -> ListNode:
    """Create linked list with optional cycle."""
    if not values:
        return
    
    head = ListNode(values[0])
    current = head
    nodes = [head]

    for val in values[1:]:
        current.next = ListNode(val)
        current = current.next
        nodes.append(current)

    # create cycle if specified
    if cycle_pos >= 0:
        current.next = nodes[cycle_pos]

    return head

def print_linked_list(head: ListNode, max_nodes: int = 10) -> None:
    """Print linked list (with cycle detection)."""
    current = head
    visited = set()
    count = 0

    while current and count < max_nodes:
        if id(current) in visited:
            print(f" -> {current.val} (CYCLE DETECTED)")
            break
        
        visited.add(id(current))
        print(f"{current.val}", end="")
        
        current = current.next
        if current:
            print(" -> ", end="")
        
        count += 1
    
    if current and count >= max_nodes:
        print(" -> ...")
    elif not current:
        print(" -> None")
    print()
```

```python
def has_cycle_verbose(head: ListNode) -> bool:
    """
    Detect cycle in linked list using Floyd's algorithm.
    """
    if not head or not head.next:
        return False
    
    slow = fast = head
    step = 0

    print("Starting cycle detection : ")
    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next
        step += 1

        print(f"Step {step}: Slow at {slow.val}, Fast at {fast.val if fast else 'None'}")
        
        if slow == fast:
            print("Cycle detected")
            return True
        
    print("No cycle detected")
```

```python
# Test 1: No cycle
print("Test 1: No cycle")
head1 = create_linked_list([1, 2, 3, 4, 5])
print("List: ", end="")
print_linked_list(head1)
has_cycle_verbose(head1)
print()

# Test 2: With cycle
print("Test 2: With cycle")
head2 = create_linked_list([3, 2, 0, -4], cycle_pos=1)
print("List: ", end="")
print_linked_list(head2)
has_cycle_verbose(head2)
print()
```

### PATTERN 4: CONTAINER/AREA PROBLEMS

#### Example: 1 - Container with Most Water

```python
def max_area_verbose(height: List[int]) -> int:
    """
    Find maximum area between two lines with detailed logging.
    """
    left, right = 0, len(height) - 1
    max_water = 0
    
    print(f"Heights: {height}")
    print("Finding maximum area:")
    
    while left < right:
        width = right - left
        current_height = min(height[left], height[right])
        current_area = current_height * width
        
        print(f"Left: {left}(h={height[left]}), Right: {right}(h={height[right]})")
        print(f"  Width: {width}, Height: {current_height}, Area: {current_area}")
        
        max_water = max(max_water, current_area)
        
        # Move pointer with smaller height
        if height[left] < height[right]:
            print(f"  Moving left pointer (smaller height)")
            left += 1
        else:
            print(f"  Moving right pointer (smaller/equal height)")
            right -= 1
        
        print(f"  Current max area: {max_water}")
    
    return max_water
```

```python
# Test Container with Most Water
print("=== CONTAINER WITH MOST WATER ===")
heights = [1, 8, 6, 2, 5, 4, 8, 3, 7]
result = max_area_verbose(heights)
print(f"Maximum area: {result}")
print()
```

### ADVANCED: SLIDING WINDOW WITH TWO POINTERS

#### Minimum Window Substring

```python
def min_window_substring_verbose(s: str, t: str) -> str:
    """
    Find minimum window in s that contains all characters of t.
    """
    need = Counter(t)
    window = defaultdict(int)
    
    left = right = 0
    valid = 0
    
    # Record minimum window
    start = 0
    min_len = float('inf')
    
    print(f"String s: '{s}'")
    print(f"Target t: '{t}'")
    print(f"Need: {dict(need)}")
    print("\nSliding window process:")
    
    while right < len(s):
        # Expand window
        char = s[right]
        right += 1
        
        print(f"\nExpanding: added '{char}' at position {right-1}")
        
        if char in need:
            window[char] += 1
            if window[char] == need[char]:
                valid += 1
                print(f"  Character '{char}' requirement satisfied")
        
        print(f"  Window: {dict(window)}, Valid: {valid}/{len(need)}")
        print(f"  Current window: '{s[left:right]}'")
        
        # Contract window when valid
        while valid == len(need):
            print(f"\n  Valid window found: '{s[left:right]}'")
            
            # Update minimum window
            if right - left < min_len:
                start = left
                min_len = right - left
                print(f"    New minimum window: '{s[start:start + min_len]}'")
            
            # Contract from left
            char = s[left]
            left += 1
            print(f"  Contracting: removed '{char}' from position {left-1}")
            
            if char in need:
                if window[char] == need[char]:
                    valid -= 1
                    print(f"    Character '{char}' requirement no longer satisfied")
                window[char] -= 1
            
            print(f"    Window after contraction: {dict(window)}, Valid: {valid}/{len(need)}")
    
    result = "" if min_len == float('inf') else s[start:start + min_len]
    print(f"\nFinal result: '{result}'")
    return result

# Test Minimum Window Substring
print("=== MINIMUM WINDOW SUBSTRING ===")
s = "ADOBECODEBANC"
t = "ABC"
result = min_window_substring_verbose(s, t)
print()
```

### PERFORMANCE COMPARISON: TWO POINTERS VS BRUTE FORCE

```python
def two_sum_brute_force(nums: List[int], target: int) -> List[int]:
    """Brute force approach for comparison."""
    for i in range(len(nums)):
        for j in range(i + 1, len(nums)):
            if nums[i] + nums[j] == target:
                return [i, j]
    return []

def performance_comparison():
    """Compare two pointers vs brute force performance."""
    print("=== PERFORMANCE COMPARISON ===")
    
    # Generate test data
    sizes = [100, 1000, 5000]
    
    for size in sizes:
        # Create sorted array for two pointers
        nums = sorted(random.sample(range(1, size * 2), size))
        target = nums[10] + nums[20]  # Ensure target exists
        
        print(f"\nArray size: {size}")
        
        # Test brute force
        start_time = time.time()
        result_bf = two_sum_brute_force(nums, target)
        bf_time = time.time() - start_time
        
        # Test two pointers
        start_time = time.time()
        result_tp = two_sum_sorted(nums, target)
        tp_time = time.time() - start_time
        
        print(f"Brute Force: {bf_time:.6f}s, Result: {result_bf}")
        print(f"Two Pointers: {tp_time:.6f}s, Result: {result_tp}")
        print(f"Speedup: {bf_time/tp_time:.2f}x")

performance_comparison()
```

### EDGE CASE TESTING FRAMEWORK

```python
def test_edge_cases():
    """Test common edge cases for two pointer problems."""
    print("=== EDGE CASE TESTING ===")
    
    # Test cases for two_sum_sorted
    test_cases = [
        ([], 0, "Empty array"),
        ([1], 1, "Single element"),
        ([1, 2], 3, "Two elements - valid"),
        ([1, 2], 4, "Two elements - invalid"),
        ([1, 1, 1, 1], 2, "All same elements"),
        ([-1, 0, 1], 0, "With negative numbers"),
        ([1, 2, 3, 4, 5], 9, "Target at end"),
        ([1, 2, 3, 4, 5], 3, "Target at beginning")
    ]
    
    print("Testing two_sum_sorted:")
    for nums, target, description in test_cases:
        result = two_sum_sorted(nums.copy(), target)
        print(f"  {description}: {nums}, target={target} -> {result}")
    
    print("\nTesting palindrome:")
    palindrome_cases = [
        ("", "Empty string"),
        ("a", "Single character"),
        ("aa", "Two same characters"),
        ("ab", "Two different characters"),
        ("racecar", "Odd length palindrome"),
        ("abccba", "Even length palindrome")
    ]
    
    for s, description in palindrome_cases:
        result = is_palindrome_verbose(s)
        print(f"  {description}: '{s}' -> {result}")

# Run edge case tests
test_edge_cases()
```

### VISUALIZATION HELPERS

```python
def visualize_two_pointers(arr: List[int], left: int, right: int, title: str = ""):
    """Visualize array with two pointers."""
    print(f"\n{title}")
    print("Array: ", end="")
    for i, val in enumerate(arr):
        if i == left and i == right:
            print(f"[{val}]", end=" ")
        elif i == left:
            print(f"L{val}", end=" ")
        elif i == right:
            print(f"R{val}", end=" ")
        else:
            print(f" {val}", end=" ")
    print()
    print("Index: ", end="")
    for i in range(len(arr)):
        print(f" {i}", end=" ")
    print()

# Example visualization
print("=== VISUALIZATION EXAMPLE ===")
arr = [1, 3, 5, 7, 9, 11]
for i in range(3):
    visualize_two_pointers(arr, i, len(arr) - 1 - i, f"Step {i + 1}")
```
